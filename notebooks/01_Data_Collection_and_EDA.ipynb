# 💳 Layer 1 & 3: Data Collection and EDA

Welcome to the first layer of the **Credit Card Fraud Detection Pipeline**.

In this notebook, we load the raw transactions and perform deep Exploratory Data Analysis (EDA) to understand class imbalance and feature distributions.

```python
import sys
import os
sys.path.append('..')
from src.preprocessing import load_data
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set beautiful style
plt.style.use('dark_background')
```

## 1. Load Data

```python
try:
    df = load_data()
    print(f'Data Loaded Successfully! Shape: {df.shape}')
except Exception as e:
    print(e)
```

## 2. Target Class Imbalance

```python
if 'df' in locals():
    plt.figure(figsize=(6,4))
    sns.countplot(data=df, x='Class', palette=['#00e5ff', '#ff007f'])
    plt.title('Extreme Class Imbalance', fontsize=16)
    plt.yscale('log')
    plt.show()
```
